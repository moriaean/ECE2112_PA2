### A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
X_mean = np.mean(X)
X_mean
```

```python
X_std = np.std(X)
X_std
```

```python
X_normalized = (X - X_mean) / X_std
X_normalized
```

#### Required Checks

```python
print("Original Array (X):")
print(X)
```

```python
print("Normalized Array (X_normalized):")
print(X_normalized)
```

```python
print("Normalized Mean:", np.mean(X_normalized))
```

```python
print("Normalized Standard Deviation:", np.std(X_normalized))
```

```python
np.save("X_normalized.npy", X_normalized)
```

### B. CUBES DIVISIBLE BY 4 PROBLEM

```python
import numpy as np
```

```python
C = (np.arange(1, 101) ** 3).reshape(10, 10)
C
```

```python
div_by_4 = C[C%4 == 0]
div_by_4
```

#### Required Checks

```python
print("Shape of C:", C.shape)
```

```python
print("Array div_by_4:")
print(div_by_4)
```

```python
print("Number of selected elements:", div_by_4.size)
```

```python
np.save("div_by_4.npy", div_by_4)
```

### C. ABOVE-MEAN SQUARES PROBLEM

```python
import numpy as np
```

```python
S = (np.arange(1, 37) ** (2)).reshape(6, 6)
S
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S > S_mean]
above_mean
```

#### Required Checks

```python
print("Array of S:")
print(S)
```

```python
print("Mean of S:")
print(S_mean)
```

```python
print("S-values above the mean:")
print(above_mean)
```

```python
print("Number of S-values above the mean:")
print(above_mean.size)
```

```python
np.save("above_mean.npy", above_mean)
```
